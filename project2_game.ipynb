This is a simple Hangman game using Python programming language. You can use this as a project to boost their programming skills and understanding logic.  

1.	The Hangman program randomly selects a secret word from a list of secret words. The random module will provide this ability, so line 1 in program imports it.

2.	The Game: Here, a random word (a fruit name) is picked up from our collection and the player gets limited chances to win the game.

3.	When a letter in that word is guessed correctly, that letter position in the word is made visible. In this way, all letters of the word are to be guessed before all the chances are over.

4.	For convenience, we have given length of word + 2 chances. For example, word to be guessed is mango, then user gets 5 + 2 = 7 chances, as mango is a five-letter word.


```python
import random

# list of secret words
fruits = ['apple','mango','banana','cherry','orange','grapes','papaya']

# Randomly select any word
word = random.choice(fruits)

# number of chances
chances = len(word) + 2

# stores guessed lettter of word
guessed_word = ["_"]*len(word)

# main code

print("WELCOME TO HANGMAN GAME : ")
print("-------------------------")

print("Guess the fruit name: ")
print("Fruit: "," ".join(guessed_word))

print(f"You have {chances} chances. \n")

# Logic
guessed_letters = []
while chances > 0 and "_" in guessed_word:
  guess_letter = input("Guess a letter : ").lower()

  if guess_letter in guessed_letters:
    print("You already guessed this letter.\n")
    continue

  guessed_letters.append(guess_letter)

  if guess_letter in word:
    for i in range(0,len(word)):
      if word[i] == guess_letter:
        guessed_word[i] = guess_letter

    print("Good Guess!")

  else:
        print("Wrong Guess!")
        chances -= 1
        print("Remaining chances: ",chances)

  print("Fruit:", " ".join(guessed_word))
  print()

if "_" not in guessed_word:
    print("Guessed word is: ",word)
    print("Congratulation.You scored well....")

else:
    print("Game Over!!! \nYou losted all chances...")
```
